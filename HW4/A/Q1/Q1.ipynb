### Libraries

```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt
```

### Reading the Image

```python
### YOUR CODE ###
img = cv2.imread("q1.png", cv2.IMREAD_GRAYSCALE)
```

### Finding Threshold

```python
# You can use cv2 methods for finding this
_, thresholded = cv2.threshold(img, 0, 255,
cv2.THRESH_BINARY | cv2.THRESH_OTSU)
```

```python
# Show the results
plt.figure(figsize=(10,4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Grayscale')

plt.subplot(1, 2, 2)
plt.imshow(thresholded, cmap='gray')
plt.title(f'Otsu Thresholding (Threshold = {_:.2f})')

plt.tight_layout()
plt.show()
```

### Finding Connected Components

```python
# You can use cv2 methods for finding them
### YOUR CODE ###
num_labels, labels_im = cv2.connectedComponents(img)
```

#### Show the result

Show each component with a random color, and write the number of components on the result image
